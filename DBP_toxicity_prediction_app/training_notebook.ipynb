# Chemical Toxicity LightGBM Model

This notebook reproduces the training workflow used by the Streamlit application.

```python
from pathlib import Path
import joblib
import pandas as pd
from lightgbm import LGBMRegressor
from sklearn.metrics import r2_score, root_mean_squared_error
```

## 1. Load the supplied training and test datasets

```python
training_data = pd.read_csv('data/training data 872026.csv')
test_data = pd.read_csv('data/Test data 872026.csv')
print('Training shape:', training_data.shape)
print('Test shape:', test_data.shape)
training_data.head()
```

## 2. Define the exact feature list

```python
feature_names = [c for c in training_data.columns if c not in ['Canonical SMILES', 'Value']]
descriptor_names = [c for c in feature_names if c not in ['Endpoint', 'Cell']]
xtrain = training_data[feature_names]
ytrain = training_data['Value'].to_numpy()
xtest = test_data[feature_names]
ytest = test_data['Value'].to_numpy()
print('Model inputs:', len(feature_names))
print('Mordred descriptors:', len(descriptor_names))
```

## 3. Train and evaluate LightGBM

```python
model = LGBMRegressor(random_state=42, verbosity=-1, n_jobs=1)
model.fit(xtrain, ytrain)
prediction = model.predict(xtest)
r2 = r2_score(ytest, prediction)
rmse = root_mean_squared_error(ytest, prediction)
print(f'Test R2: {r2:.4f}')
print(f'Test RMSE: {rmse:.4f}')
```
